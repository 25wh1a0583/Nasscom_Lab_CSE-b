Funtion without arguments


```python
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

Function with arguments

```python
def greet_user(name):
    print("Hello," ,name,"!")
greet_user("Alice")
greet_user("Bob")
```

Function with default parameters

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

Function with *args

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

6. USING **kwargs (VARIABLE NUMBER OF KEYWORD ARGUMENTS)

 *kwargs lets you pass any number of keyword arguments as a dictionary




```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")
print_info(name="Alice", age=25, city="Delhi")
```

Combination of *args and **kwargs

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

Nested Function

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

✅ Summary Table

| Feature        | Use Case Example                       |
| -------------- | -------------------------------------- |
| Basic Function | `def greet():`                         |
| With Args      | `def greet(name):`                     |
| With Default   | `def greet(name="Guest"):`             |
| Return Value   | `return result`                        |
| `*args`        | `def func(*args):` → Tuple of args     |
| `**kwargs`     | `def func(**kwargs):` → Dict of kwargs |
| Nested         | One function calling another           |


```python
from functools import reduce
numbers = [1, 2, 3, 4, 5, 6]
squares = list(map(lambda x: x**2, numbers))
print(squares)
odd = list(filter(lambda x: x % 2 != 0, squares))
print(odd)
total = reduce(lambda x,y:x+y,squares)
print(total)
```

```python
dict={'a':1,'b':2,'c':3}
for key,value in dict.items():
  print(key,value)
squares={key: value**2 for key, value in dict.items()}
print(squares)
filter={key: value for key, value in squares.items() if value % 2 != 0}
print(filter)
total=reduce(lambda x,y:x+y,squares.values())
print(total)
```
