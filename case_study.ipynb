```python
import numpy as np
time = [0, 1, 2, 3, 4, 5]
temperature = [32, 35, 38, 40, 42, 45]
voltage = [400, 395, 390, 385, 380, 375]
current = [20, 22, 25, 28, 30, 32]
charge = [95, 88, 80, 70, 60, 48]

print("BATTERY SENSOR READINGS")

for i in range(len(time)):
    print(
        "Time:", time[i], "h",
        "| Temperature:", temperature[i], "°C",
        "| Voltage:", voltage[i], "V",
        "| Current:", current[i], "A",
        "| Charge:", charge[i], "%"
    )


print("\nTEMPERATURE STATUS")

for t in temperature:
    if t > 40:
        print(t, "°C -> High Temperature Alert")
    else:
        print(t, "°C -> Temperature Normal")


print("\nBATTERY STATUS")

for c in charge:
    if c < 50:
        print(c, "% -> Low Battery Alert")
    else:
        print(c, "% -> Battery Charge Normal")



def battery_status(charge):
    if charge > 70:
        return "High"
    elif charge >= 50 and charge <= 70:
        return "Medium"
    else:
        return "Low"


print("\nBATTERY STATUS USING FUNCTION")

for c in charge:
    print(c, "% ->", battery_status(c))

# TASK 2: NUMPY ANALYSIs

print("NUMPY ANALYSIS")

D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])

print("\nBattery Data:")
print(D)

print("\nShape of array:", D.shape)

print("Number of rows:", D.shape[0])
print("Number of columns:", D.shape[1])

average = np.mean(D, axis=0)

print("\nAverage Temperature:", average[0])
print("Average Voltage:", average[1])
print("Average Current:", average[2])
print("Average Charge:", average[3])

print("\nMaximum Temperature:", np.max(D[:, 0]))
print("Minimum Temperature:", np.min(D[:, 0]))

voltage_column = D[:, 1]

print("\nVoltage Column:")
print(voltage_column)

high_temperature = D[D[:, 0] > 40]

print("\nHigh Temperature Readings:")
print(high_temperature)

low_battery = D[D[:, 3] < 50]

print("\nLow Battery Readings:")
print(low_battery)

print("LINEAR ALGEBRA")

A = np.array([
    [2, 1],
    [1, 3]
])


X = np.array([
    [40],
    [25]
])

Y = np.dot(A, X)

print("\nMatrix A:")
print(A)

print("\nVector X:")
print(X)

print("\nY = AX:")
print(Y)

print("\nTranspose of A:")
print(A.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))


print("\nInverse of A:")
print(np.linalg.inv(A))

B = np.array([
    [100],
    [120]
])

solution = np.linalg.solve(A, B)

print("\nSolution of AX = B:")
print("x =", solution[0][0])
print("y =", solution[1][0])

# TASK 4: CALCULUS

print("CALCULUS ANALYSIS")

def charge_function(t):
    return 100 - 2 * t - 0.5 * t**2

def discharge_rate(t):
    return -2 - t

print("\nC(t) = 100 - 2t - 0.5t²")
print("dC/dt = -2 - t")

print("\nAt t = 2 hours:")
print("Discharge rate =", discharge_rate(2), "% per hour")

print("\nAt t = 5 hours:")
print("Discharge rate =", discharge_rate(5), "% per hour")


print("FINAL ANALYSIS")

for i in range(len(time)):
    if temperature[i] > 40:
        print("\nTemperature first exceeds 40°C at:", time[i], "hours")
        break


charge_loss = charge[0] - charge[-1]

print("Total Charge Reduction:", charge_loss, "%")

print("Average Temperature:", average[0])
print("Average Voltage:", average[1])
print("Average Current:", average[2])
print("Average Charge:", average[3])

print("\nTransformed Output Y:")
print(Y)

print("\nSystem Solution:")
print("x =", solution[0][0])
print("y =", solution[1][0])

print("\nBattery discharge rate increases with time.")
print("The negative value of dC/dt shows that the battery is losing charge.")

print("\nFINAL DECISION: MONITOR")
print("Reason: Temperature exceeds 40°C and battery charge drops below 50%.")
```
