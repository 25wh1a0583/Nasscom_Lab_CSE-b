```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
print("Number of students:", len(df))
print("\nColumn names:")
print(df.columns)
print("\nData types:")
print(df.dtypes)
print("\nFirst five records:")
print(df.head())
print("\nLast five records:")
print(df.tail())
print("\nComplete Dataset:")
print(df)
print("\nAverage values:")
print(df[["Study_Hours", "Attendance", "Assignment_Score","Previous_Score", "Final_Score"]].mean())
highest = df.loc[df["Final_Score"].idxmax()]
lowest = df.loc[df["Final_Score"].idxmin()]
print("\nStudent with highest Final Score:")
print(highest)
print("\nStudent with lowest Final Score:")
print(lowest)
print("\nStudents with Attendance > 80%:")
print(df[df["Attendance"] > 80])
print("\nStudents with Final Score > 80:")
print(df[df["Final_Score"] > 80])
print("\nStatistical Summary:")
print(df.describe())
final_scores = df["Final_Score"].to_numpy()
print("\nFinal Scores:", final_scores)
print("Mean:", np.mean(final_scores))
print("Median:", np.median(final_scores))
print("Maximum:", np.max(final_scores))
print("Minimum:", np.min(final_scores))
print("Standard Deviation:", np.std(final_scores))
plt.figure(figsize=(7, 5))
plt.scatter(df["Study_Hours"], df["Final_Score"], color="blue")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.grid(True)
plt.show()
plt.figure(figsize=(7, 5))
plt.scatter(df["Attendance"], df["Final_Score"], color="green")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.grid(True)
plt.show()
plt.figure(figsize=(7, 5))
plt.hist(df["Final_Score"], bins=5, color="orange", edgecolor="black")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Scores")
plt.grid(axis="y")
plt.show()
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Final_Score"], color="purple")
plt.xlabel("Student")
plt.ylabel("Final Score")
plt.title("Student-wise Final Scores")
plt.show()
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
print("\nMatrix X:")
print(X)
print("\nShape of X:")
print(X.shape)
print("\nTranspose X^T:")
print(X.T)
print("\nX^T X:")
print(X.T @ X)
study_hours = df["Study_Hours"].values
final_score = df["Final_Score"].values

print("\nDot product of Study Hours and Final Score:")
print(np.dot(study_hours, final_score))
def F(x):
    return 40 + 5*x - 0.2*x**2
def dF(x):
    return 5 - 0.4*x
print("\nStudy Hours Model:")
for x in range(1, 11):
    print("x =", x, "F(x) =", F(x), "F'(x) =", dF(x))
x_critical = 5 / 0.4
print("\nCritical point:", x_critical)
print("F(critical point):", F(x_critical))
integral_value = 2.5 * (5**2) - (0.2/3) * (5**3)
print("\nAnalytical integral:", integral_value)
x_values = np.linspace(0, 5, 1000)
L_values = 5*x_values - 0.2*x_values**2
numerical_integral = np.trapezoid(L_values, x_values)
print("Numerical integral:", numerical_integral)
correlation_matrix = df.corr(numeric_only=True)
print("\nCorrelation Matrix:")
print(correlation_matrix)
print("\nCorrelations with Final Score:")
print(correlation_matrix["Final_Score"].sort_values(ascending=False))
plt.figure(figsize=(9, 7))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values
X_model = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X_model.T @ X_model) @ X_model.T @ Y
print("\nPrediction Model Coefficients:")
print(beta)
print("\nModel:")
print(
    f"Final Score = {beta[0]:.4f} "
    f"+ ({beta[1]:.4f}) Study Hours "
    f"+ ({beta[2]:.4f}) Attendance "
    f"+ ({beta[3]:.4f}) Assignment Score"
)
```
